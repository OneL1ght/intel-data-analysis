```python
# task 1

def get_crypt(msg):
    msg = list(msg)
    answer = ""
    letter = msg.pop(0)
    counter = 1
    while msg:
        curr_letter = msg.pop(0)
        if curr_letter == letter:
            counter += 1
        else:
            answer += f'{counter}{letter}' if counter > 1 else letter
            counter = 1
            letter = curr_letter
    answer += f'{counter}{letter}' if counter > 1 else letter
    return answer

print(get_crypt("AAACBBBBBDDADDD") == "3AC5B2DA3D")
print(get_crypt("GGGGGGTTTYUU") == "6G3TY2U")
print(get_crypt("D") == "D")
```

```python
# task 2. password validation

def is_valid_password(password):
    if len(password) < 12:
        return False
    
    has_upper = any(c.isupper() for c in password)
    has_lower = any(c.islower() for c in password)
    has_digit = any(c.isdigit() for c in password)
    has_special = any(c in '!@#$%&*+' for c in password)

    return has_upper and has_lower and has_digit and has_special

n = int(input())
results = []

for _ in range(n):
    password = input().strip()
    if is_valid_password(password):
        results.append("Valid")
    else:
        results.append("Invalid")

for result in results:
    print(result)
```

```python
# task 3. quartiles

def get_median(counts):
    n = len(counts)
    if n % 2 == 0:
        return (counts[n//2] + counts[n//2 - 1]) / 2
    return counts[n//2]

def get_quartiles(counts):
    counts, n = sorted(counts), len(counts)
    q2 = get_median(counts)
    if n % 2 == 0:
        fh, lh = counts[:n//2], counts[n//2:]
    else:
        fh, lh = counts[:n//2], counts[n//2+1:]
    q1, q3 = get_median(fh), get_median(lh)
    return q1, q2, q3

# n = int(input())
counts = [25, 24, 19, 99, 18, 17, 0, 21, 26]

# for _ in range(n):
#     counts.append(int(input()))

q1, q2, q3 = get_quartiles(counts)
iqr = q3 - q1
outliars = sum(1 for i in counts if (q1 - 1.5 * iqr) > i or i > (q3 + 1.5 * iqr))
print(outliars)
```

```python
# task 6. mbox
import re
from collections import Counter

with open('mbox.txt', 'r') as f:
    text = f.read().split('\n')
mails = [re.search(r'^From (.*?) ', s) for s in text]
mails = [match.group(1) for match in mails if match is not None]
mails_counts = Counter(mails)
print(mails_counts.most_common(1))
```
