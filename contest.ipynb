```python
# t 1

def solution(arr):
    first, last = 0, 0
    is_even_fun = lambda x: x % 2 == 0
    is_even = is_even_fun(arr[0])
    for i in range(1, len(arr)):
        last = i
        is_curr_even = is_even_fun(arr[i])
        if is_curr_even != is_even:
            dist = last - first
            for j in range(first, last):
                arr[j] = dist
            first = i
        is_even = is_curr_even

    last += 1
    dist = last - first
    for j in range(first, last):
        arr[j] = dist
    return arr

print(solution([2, 1, 3, 2]) == [1, 2, 2, 1])
print(solution([2, 1, 3, 9, 2]) == [1, 3, 3, 3, 1])
print(solution([2, 1, 3, 9, 2, 4, 100]) == [1, 3, 3, 3, 3, 3, 3])
print(solution([2]) == [1])
```

```python
# t 2

def get_longest_right_substring(s: str):
    """right substring - start with nulls and ended with same amount of nulls"""
    nulls, ones = 0, 0
    max_count = 0
    for i in range(len(s)):
        if s[i] == '0':
            if i-1 > 0 and s[i-1] == '1':
                nulls, ones = 0, 0
            nulls += 1
        elif s[i] == '1':
            ones += 1
        if nulls == ones:
            max_count = max(max_count, nulls + ones)
    return max_count

print(get_longest_right_substring("010001111001010001") == 6)
print(get_longest_right_substring("01") == 2)
print(get_longest_right_substring("1") == 0)
print(get_longest_right_substring("0001111111000000000111111111010101") == 18)
```
